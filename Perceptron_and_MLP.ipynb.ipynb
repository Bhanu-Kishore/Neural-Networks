# Simple perceptron
---



*   Trained On: Iris Dataset
*   Activation Function: Step



```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
iris=load_iris()
inputs = iris.data[:,:2]
targets = iris.target
mask= targets!=2
inputs=inputs[mask]
targets=targets[mask]
inputs_train, inputs_test, targets_train, targets_test=train_test_split(inputs,targets,test_size=0.3,random_state=42)
weights=np.array([1,1])
def step(x):
  return 0 if x<0 else 1
def learning(wei,inp,e,l=0.2):
  return wei+(e*l*inp)
epochs=int(input("Enter the number of eppochs: "))
for epoch in range(epochs):
  for inp in range(len(inputs_train)):
    input_p=inputs_train[inp]
    target=targets_train[inp]
    output=np.dot(input_p,weights)
    pred=step(output)
    error=target-pred
    if error!=0:
      weights=learning(weights,input_p,error)
print("Current weights after training:",weights)
correct=0
for i in range(len(inputs_test)):
  input_t=inputs_test[i]
  target_t=targets_test[i]
  o=np.dot(input_t,weights)
  pred=step(o)
  if pred==target_t:
    correct+=1
print("Accuracy: ",(correct/len(targets_test)))
```

# Multi Layer Perceptron
---





*   Hidden layers : 1
*   Hidden Neurons: 2
*   Trained On: XOR
*   Activation Fuunction: Sigmoid


```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])
w1=np.random.randn(2,2)
w2=np.random.randn(2,1)
b1=np.zeros(2)
b_final=0.0
lr=1

#---------- Sigmoid Activation Function ----------
def sig(x):
  return 1/(1+np.exp(-x))

#Epochs
ep=1000000

#---------------------------------------------------
#               Training                           |
#---------------------------------------------------
for i in range(ep):
  for j in range(len(X)):
    inp=X[j]


    tar=y[j][0]

  #Calculate weigthed sum at H1

    z1=np.dot(inp,w1[0]) + b1[0]

  #Calculate weigthed sum at H2

    z2=np.dot(inp,w1[1]) + b1[1]


    h1=sig(z1)


    h2=sig(z2)

  #H1 and H2 are inouts for the input neuron

    inp_final=np.array([h1,h2])

  #Calculate weighted sum at the Output neuron

    z_final=np.dot(inp_final,w2).item() + b_final


    out=sig(z_final)


    loss=1/2*((out-tar)**2)


    r'''
      w11      w13     (sig)
    1----z1--H1----      |
   w12\ /           \    v
       \              z--O---L
   w21/ \           /
    2----z2--H2-----
      w22      w23
    '''
    dl_dout=out-tar     #l=(1/2)*((out-tar)**2)


    dout_dz=out*(1-out) #out=1/(1+(e^(-z)))


    dz_dw13=h1          #z=H1*w13+H2*w23

  #calculate gradient for w13

    dl_dw13=dz_dw13*dout_dz*dl_dout


    dz_dw23=h2#z=H1*w13+H2*w23

  #Calculate gradient for W23

    dl_dw23=dz_dw23*dout_dz*dl_dout


    dz_dh1=w2[0,0]#z=H1*w13+H2*w23


    dl_dh1=dz_dh1*dout_dz*dl_dout


    dz_dh2=w2[1,0]#z=H1*w13+H2*w23


    dl_dh2=dz_dh2*dout_dz*dl_dout


    dh1_dz1=h1*(1-h1)#H1=1/(1+e^(-z1))


    dl_dz1=dh1_dz1*dl_dh1


    dh2_dz2=h2*(1-h2)#H2=1/(1+e^(-z2))


    dl_dz2=dh2_dz2*dl_dh2

    #calculate gradient for W11


    dz1_dw11=inp[0]#z1=w11[x1]+w21[x2]


    dl_dw11=dz1_dw11*dl_dz1

    #Calculate gradient for w21


    dz1_dw21=inp[1]


    dl_dw21=dz1_dw21*dl_dz1

    #Calculate gradients for Bias b1

    dz1_db1 = 1

    dl_db1 = dz1_db1*dl_dz1

    #Calculate gradient for Bias b2

    dz2_db2 = 1

    dl_db2 = dz2_db2 * dl_dz2

    #Calculate gradient for b_final

    dout_dbfinal = 1

    dl_dbfinal = dl_dout * dout_dz

    #Calculate gradient for w12

    dz2_dw12 = inp[0]   #z2 = w12[x1] + w22[x2]

    dl_dw12 = dz2_dw12*dl_dz2

    #Calculate gradient for w22

    dz2_dw22 = inp[1]

    dl_dw22 = dz2_dw22 * dl_dz2

    #Update the weights W13 and W23

    w2[0,0] = w2[0,0] - lr * dl_dw13   #w2[0]=w13
    w2[1,0] = w2[1,0] - lr * dl_dw23   #w2[1]=w23

    #Update the weight w11 and W21

    w1[0][0] = w1[0][0]-lr*dl_dw11   #W11
    w1[0][1] = w1[0][1]-lr*dl_dw21   #W21

    #update the weights W12 and W22

    w1[1][0] = w1[1][0]-lr*dl_dw12   #W12
    w1[1][1] = w1[1][1]-lr*dl_dw22   #W22

    #Update the biases b1 and b2

    b1[0] = b1[0] - lr * dl_db1
    b1[1] = b1[1] - lr * dl_db2

    #Update the final output bias b_final

    b_final = b_final - lr * dl_dbfinal
#--------------- Training Completed---------------------#

#-------Testing-----------#

for j in range(len(X)):

    inp = X[j]
    tar = y[j]

    # Hidden layer
    z1 = np.dot(inp, w1[0]) + b1[0]
    z2 = np.dot(inp, w1[1]) + b1[1]

    h1 = sig(z1)
    h2 = sig(z2)

    # Output layer
    inp_final = np.array([h1, h2])

    z_final = np.dot(inp_final, w2).item() + b_final

    out = sig(z_final)

    # Convert sigmoid output to class
    pred = 1 if out >= 0.5 else 0

    print("Input:", inp,
          "Target:", tar[0],
          "Output:", out,
          "Prediction:", pred)
```
